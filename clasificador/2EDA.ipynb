```python
import pickle
import pandas as pd
import seaborn as sns
```

```python
with open('data.pkl','rb') as f:
    data = pickle.load(f)
```

```python
df_dogs = pd.DataFrame(
    {
    'label':'dog',
    'high': [img.shape[0] for img in data['dogs']],
    'widh': [img.shape[1] for img in data['dogs']]
    },
)
```

```python
df_cats = pd.DataFrame(
    {
    'label':'cat',
    'high': [img.shape[0] for img in data['cats']],
    'widh': [img.shape[1] for img in data['cats']]
    },
)
```

### EDA cats

```python
df = pd.concat( [df_dogs,df_cats], ignore_index=True)
```

```python
df.info()
```

```python
df.describe().T
```

```python
sns.histplot(df,x='widh',bins=6,stat='percent')
#sns.histplot(df,x='high',bins=6,stat='percent')
```
