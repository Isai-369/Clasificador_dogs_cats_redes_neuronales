```python
import cv2
import matplotlib.pyplot as plt
import os
import pickle
```

```python
PATH_CATS ="C:/Users/anonimo/Documents/Dataset/petImages/cat"
PATH_DOGS ="C:/Users/anonimo/Documents/Dataset/petImages/dog"
```

```python
def get_data_from_path(path):

    if os.path.exists(path):     

        list_images = []          

        for imagen in os.listdir(path):

            img = cv2.imread( os.path.join(path,imagen) )            
            
            if img is not None:
                img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
                list_images.append(img)
        
        return list_images
    else:
        print("no")

def save_data(data,path):
    try:
        with open(path,'wb') as f:
            pickle.dump(data,f)
    except Exception as e:
        print(e)
```

```python
cats = get_data_from_path(PATH_CATS)
```

```python
dogs = get_data_from_path(PATH_DOGS)
```

```python
data = {
    'dogs':dogs,
    'cats':cats
} 

save_data(data,'data.pkl')
```
