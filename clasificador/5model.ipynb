```python
import pickle
import numpy as np
```

```python
with open('data_train_test.pkl','rb') as f:
    df = pickle.load(f)
    if df:
        X_train, X_test, y_train, y_test = df['x_train'] , df['x_test'] , df['y_train'], df['y_test']
```

```python
from keras import layers, models,optimizers

def crear_modelo_cnn( input_shape = (224, 224, 3) ):
    
    model = models.Sequential()
    
    # 1. Primera capa convolucional + Pooling
    # Extrae características básicas (bordes, contrastes)
    model.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=input_shape))
    model.add(layers.MaxPooling2D((2, 2)))
    
    # 2. Segunda capa convolucional + Pooling
    # Combina las características anteriores en patrones más complejos (texturas)
    model.add(layers.Conv2D(64, (3, 3), activation='relu'))
    model.add(layers.MaxPooling2D((2, 2)))
    
    # 3. Tercera capa convolucional + Pooling
    # Detecta formas y estructuras geométricas más abstractas
    model.add(layers.Conv2D(128, (3, 3), activation='relu'))
    model.add(layers.MaxPooling2D((2, 2)))
    
    # 4. Capa de aplanado (Flatten)
    # Transforma la matriz 2D de características en un vector 1D
    model.add(layers.Flatten())
    
    # 5. Capa densa (Fully Connected) para la interpretación de las características
    model.add(layers.Dense(128, activation='relu'))
    # Dropout para evitar el sobreajuste (overfitting)
    model.add(layers.Dropout(0.5))
    
    # 6. Capa de salida
    # Usamos 1 neurona con 'sigmoid' porque es una clasificación binaria (0 o 1)
    model.add(layers.Dense(1, activation='sigmoid'))
    
    return model

# --- Construir y Compilar el Modelo ---
model = crear_modelo_cnn()

mi_optimizador = optimizers.Adam(learning_rate=0.001)

model.compile(
    optimizer = mi_optimizador,              # Optimizador estándar y eficiente
    loss='binary_crossentropy',            # Función de pérdida ideal para clasificación binaria
    metrics=['accuracy']                   # Métrica para evaluar el rendimiento
)

# Ver el resumen de la arquitectura y la cantidad de parámetros
model.summary()
```

```python
print("Iniciando el entrenamiento del modelo convolucional...")

history = model.fit(
    X_train,
    y_train,
    epochs = 20,
    batch_size = 32,                  # Es buena práctica definir el tamaño del lote (por defecto es 32)
    validation_data = (X_test, y_test), 
    verbose = 1
)
```

```python
# Guarda el modelo completo al finalizar el entrenamiento
model.save('clasificador_pets_cnn.keras')
print("¡Modelo guardado con éxito en formato .keras!")
```

```python
# Guardar el historial de entrenamiento por separado
def save_data(data,path):
    try:
        with open(path,'wb') as f:
            pickle.dump(data,f)
    except Exception as e:
        print(e)
```

```python
save_data(history.history, 'historial_entrenamiento.pkl')
```
