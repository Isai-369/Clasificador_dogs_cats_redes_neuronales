```python
from keras.models import load_model
import cv2 
import os
import matplotlib.pyplot as plt
import numpy as np
```

```python
model = load_model("clasificador_pets_cnn.keras")
```

```python
def prediction(path):
    try:
        if os.path.exists(path):            
            img = cv2.imread(path)
            img = cv2.resize(img,(224,224))
            img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
            
            img = img.astype('float32')/255.0
            plt.imshow(img)
            img = np.expand_dims(img,axis=0)

            predict = model.predict(img,verbose=0)[0][0]
            
            if predict > 0.5 :                
                print(f"Perro {predict:0.2f}")
            else:                 
                print(f"Gato {1.0-predict:0.2f}")            
        else:
            print("No encontrada")
    except Exception as e:
        print(e)
    
```
